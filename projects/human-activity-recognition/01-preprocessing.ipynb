```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load datasets
train_df = pd.read_csv('data/metadata.csv')
test_df = pd.read_csv('data/metadata_test.csv')
kaggle_df = pd.read_csv('data/metadata_kaggle.csv')  # No target column

# Drop identifier column
train_df.drop(columns=["user_snippet"], inplace=True)
test_df.drop(columns=["user_snippet"], inplace=True)

X_train = train_df.drop(columns=["activity"])
y_train = train_df["activity"]
X_test = test_df.drop(columns=["activity"])
y_test = test_df["activity"]
X_kaggle = kaggle_df.drop(columns=["user_snippet"])
```

```python
import os
os.makedirs('data-prep', exist_ok=True)
X_train.to_csv('data-prep/X_train.csv', index=False)
y_train.to_csv('data-prep/y_train.csv', index=False)
X_test.to_csv('data-prep/X_test.csv', index=False)
y_test.to_csv('data-prep/y_test.csv', index=False)
X_kaggle.to_csv('data-prep/X_kaggle.csv', index=False)
```

```python
missing_train = train_df.isnull().sum()
missing_test = test_df.isnull().sum()
missing_kaggle = kaggle_df.isnull().sum()

print("Missing values in train_df:")
print(missing_train[missing_train != 0])

print("\nMissing values in test_df:")
print(missing_test[missing_test != 0])

print("\nMissing values in kaggle_df:")
print(missing_kaggle[missing_kaggle != 0])
```
